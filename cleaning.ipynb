## Cleaning prices_monthly.csv

We will first start by downloading the prices_monthly.csv file into this notebook.

```python
import urllib.request

# The prices_monthly.csv file is downloaded into the notebook
urllib.request.urlretrieve("https://raw.githubusercontent.com/xiendou/team24_project_lab/refs/heads/main/data/pre_processed/prices_monthly.csv", "prices_monthly.csv")
print("Downloaded prices_monthly.csv")
```

Pandas is needed to allow us to to open and control the table.

```python
import pandas as pd

# The database can now be accessed using the variable identifier prices_monthly
prices_monthly = pd.read_csv("prices_monthly.csv")
```

It is important to note what kind of data prices_monthly stores, which is done by identifying what each row represents.

Each row represents the prices of different commodities on the first day of a specific month.

As we are only interested in the cocoa commodity, we will focus on its column as required.

We will now gain insights into certain numerical information in the table, such as the number of entries per column.

```python
# Format: (number of rows, number of columns) for the prices_monthly table
prices_monthly.shape
```

```python
# Display the type of data in each column
prices_monthly.info()
```

As seen above, all the commodities, except for soybeans, is represented by a float64 data type, i.e. numbers with decimal places.

The data type for soybeans is an integer due it being a much higher-priced commodity. This choice was made by The World Bank in their Pink Sheet. Nonetheless, this is does not affect our question as we are only interested in the cocoa column.

Also, each column has 442 non-null entries, which matches the 442 entries in the table. Hence, there are no blank entries.

We will now find some statistical insights into the cocoa column.

```python
# Gives some statistical insight into the 'cocoa' column
prices_monthly["cocoa"].describe()
```

After searching through the table above for erroneous data, none has appeared.

For example, the minimum is above 0, as would be expected for a commodity's price.

The count of 442 entries matches the non-null count for the cocoa column when prices_monthly.info() was called.

It is important that there are no blank entries in both the months column and the cocoa column. Although we have seen this already from the function call prices_monthly.info(), we can focus on this specifically using the following function:

```python
# Gives the number of null entries per column
prices_monthly.isna().sum()
```

As seen above, none of the columns have any null (i.e. empty) entries, including the cocoa column.

It is also important that there are no duplicated rows or months, as they may skew the results when analysing.

```python
# Checks for any duplicated rows
prices_monthly.duplicated().sum()
```

```python
# Checks for any duplicated month entries
prices_monthly["month"].duplicated().any()
```

The number of duplicated rows is 0. Hence, we do not need to drop any at the moment.
And, there are no duplicated months.

It is worth noting that this is for rows and/or entries that match exactly, hence entry mistakes or inconsistent date formats are not picked up.

A potential issue is that the date formats are simply stored as text, and hence not recognised as a date object. This is demonstrated using the function call below:

```python
prices_monthly["month"].describe()
```

If the month column was treated as a proper date data type, values such as the minimum, maximum, median, etc. would be displayed, but they are not.

We can convert it to the desired date type using the code below:

```python
# Turns the text into real dates. %Y-%m-%d means year/month/day
prices_monthly["month"] = pd.to_datetime(prices_monthly["month"], format = "%Y-%m-%d")

prices_monthly["month"].describe()
```

As shown above, it has worked as we now have certain numbers such as the mean, minimum, median, maximum, etc., which we did not have before.

We can also now spot that there are no erroneous dates with regards to a date being too distant in either the past or future.

## Cleaning weather_monthly.csv

A similar procedure to the one above will be followed to clean weather_monthly.csv.

We will first start by downloading the weather_monthly.csv file into this notebook.

```python
# The weather_monthly.csv file is downloaded into the notebook
urllib.request.urlretrieve("http://raw.githubusercontent.com/xiendou/team24_project_lab/refs/heads/main/data/pre_processed/weather_monthly.csv", "weather_monthly.csv")

# Pandas is used to control the table as a variable
weather_monthly = pd.read_csv("weather_monthly.csv")

print("Downloaded weather_monthly.csv")
```

It is important to note what kind of data weather_monthly stores, which is done by identifying what each row represents.

Each row represents the atmospheric data of a specific region on Earth on the first day of a specific month.

As we are only interested in the cocoa commodity, we will focus only on the rows whose region is where cocoa is planted.

To do this, we need a list of the unique entries in the region column.

```python
weather_monthly["region"].value_counts()
```

We already know that out of the regions in the region column, ivory_coast_daloa and ghana_kumasi are the two largest global producers of cocoa.

Hence, we will only focus on them, and remove any row that does not have either of them as its region.

```python
# Before deleting the unwanted rows, we will find the dimension of the table
print(weather_monthly.shape)

# Keeps rows whose entry in the region column is either "ivory_coast_daloa" or "ghana_kumasi"
weather_monthly = weather_monthly[(weather_monthly["region"] == "ivory_coast_daloa") | (weather_monthly["region"] == "ghana_kumasi")]
```

We will now gain insights into certain numerical information in the table, such as the number of entries per column.

```python
# Format: (number of rows, number of columns) for the prices_monthly table
weather_monthly.shape
```

```python
# Display the type of data in each column
weather_monthly.info()
```

As shown above, the total number of rows in the table has been reduced from 4,420 rows to 884 rows.

It can also be noted that none of the columns have an empty entry, as each non-null count is equal to the total number of rows.

It is also worth noting that the month column is stored as an object, which will later have to be turned into a date data type. All the other columns store an appropriate data type.

We will now find some statistical insights into each number column.

```python
weather_monthly.describe()
```

After searching through the table above for erroneous data, none has appeared, as the mean, minimum, lower quartile, median, upper quartile and maximum of each column respect the range of each column (e.g. the minimum of precip_mm is non-negative, as would be expected).

The count of 884 entries matches the non-null count for each of the columns, which confirms to us once again that there are no empty entries.

We can check once again for empty entries using the following function call:

```python
# Gives the number of null entries per column
weather_monthly.isna().sum()
```

Once again, it is confirmed that there are no empty entries.
